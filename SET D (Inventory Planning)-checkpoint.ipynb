```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Inventory_DF = pd.read_csv("D:\OneDrive\Desktop\Interview Prepe\Practical Exam\Inventory Planing\Inventory.csv")
Item_DF = pd.read_csv("D:\OneDrive\Desktop\Interview Prepe\Practical Exam\Inventory Planing\Items.csv")

Inventory_DF
```

```python
Item_DF
```

```python
print(f"Rows before remove Duplicates {len(Inventory_DF)}")
Remove_Duplicates = Inventory_DF.drop_duplicates()
print(f"Rows after remove Duplicates {len(Remove_Duplicates)}")
```

```python
Merged = Inventory_DF.merge(Item_DF , on = "item_id" , how = "left")
Merged
```

```python
Merged["Shortage_Units"] = (Merged["reorder_level"] - Merged["stock"]).clip(lower = 0) 
Merged
```

```python
Category_Summary = (Merged.groupby("category").agg(Total_Shortage_Units = ("Shortage_Units" , "sum") , 
                                                   Snapshot = ("Shortage_Units" , "count") , 
                                                   shortage_snapshots = ("Shortage_Units" , lambda x : (x > 0).sum())))
Category_Summary
```

```python
Category_Summary["Shortage_Incidence_Rate"] = (Category_Summary["shortage_snapshots"] / Category_Summary["Snapshot"]) * 100
Category_Summary = Category_Summary.round(2)
Category_Summary
```

```python
Item_Summary = Merged.groupby(["item_id" , "item"])["Shortage_Units"].sum().reset_index().sort_values("Shortage_Units" , ascending = False)
Item_Summary
```

```python
Top_Item = Item_Summary.iloc[0]
Top_Item_Shortage = Top_Item["Shortage_Units"]
Overall_Shortage = Merged["Shortage_Units"].sum()
Top_Item_Share = (Top_Item_Shortage / Overall_Shortage) * 100

print(f"Top Item : {Top_Item['item']}")
print(f"Top Item Shortage : {Top_Item_Shortage}")
print(f"Share of Total Shortage : {round(Top_Item_Share, 2)} %")
```

```python
Month_Order = ["Jan" , "Feb" , "Mar"]

Monthly_AVG = Merged.groupby("month")["Shortage_Units"].sum().reindex(["Jan" , "Feb" , "Mar"])


plt.figure(figsize = (8 , 5))
plt.plot(Monthly_AVG.index , Monthly_AVG.values , marker = "o")
plt.title("Monthly Total Shortage Units")
plt.xlabel("Months")
plt.ylabel("Total Shortage Units")

plt.show()
```
