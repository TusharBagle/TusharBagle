# Remove the Duplicates 

```python
print(f"Total Rows before Removing Duplicates : {len(Deliveries_DF)}")
New_Deliveries_DF = Deliveries_DF.drop_duplicates()
print(f"Total Rows after Removing Duplicates : {len(New_Deliveries_DF)}")
```

# Merged the Both tables

```python
Merged_Tables = New_Deliveries_DF.merge(Restaurant_DF , on = "restaurant_id" , how = "left")
Merged_Tables
```

# Calculate Late Flag

```python
Merged_Tables["Late_Flag"] = (New_Deliveries_DF["delivery_minutes"] > 35).astype(int)
Merged_Tables
```

# Grouped Summary By Curisine

```python
Total_deliveries = len(Merged_Tables)
Total_Late_Flag = int((Merged_Tables["Late_Flag"] == 1).sum())
Late_Deliveries = (Total_Late_Flag / Total_deliveries) * 100
Rate_Late_Deliveries = round(Late_Deliveries , 2)

print(f"Total Deliveries : {Total_deliveries}")
print(f"Total Late Flags : {Total_Late_Flag}")
print(f"The Late Deliveries Rate is : {Rate_Late_Deliveries} %")
```

# Cuisine Analysis

```python
cuisine_summary = (
    Merged_Tables
    .groupby("cuisine")
    .agg(
        Total_deliveries=("delivery_id", "count"),
        Late_Deliveries=("Late_Flag", "sum")
    )
    .reset_index()
)

cuisine_summary["Rate_Late_Deliveries"] = (
    cuisine_summary["Late_Deliveries"]
    / cuisine_summary["Total_deliveries"]
    * 100
)

print(cuisine_summary)
```

# Identify highest cuisine

```python
highest = cuisine_summary.loc[
    cuisine_summary["Rate_Late_Deliveries"].idxmax()
]

print(
    f"Highest late-delivery cuisine: "
    f"{highest['cuisine']}"
)

print(
    f"Late deliveries: "
    f"{highest['Late_Deliveries']} / "
    f"{highest['Total_deliveries']}"
)

print(
    f"Late-delivery rate: "
    f"{highest['Rate_Late_Deliveries']:.2f}%"
)
```

# Monthly analysis

```python
Month_Order = ["Jan", "Feb", "Mar"]

Monthly_AVG = Merged_Tables.groupby("month", observed = True)["delivery_minutes"].mean().reset_index()


plt.figure(figsize=(8, 5))

plt.plot(Monthly_AVG["month"] , Monthly_AVG["delivery_minutes"] , marker = "o")

plt.title("Monthly Average Delivery Time")
plt.xlabel("Month")
plt.ylabel("Average Delivery Time (Minutes)")

plt.show()
```
