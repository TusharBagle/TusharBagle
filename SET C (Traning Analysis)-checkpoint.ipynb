```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Asssesment_DF = pd.read_csv("D:\\OneDrive\\Desktop\\Interview Prepe\\Practical Exam\\Tranning Analysis\\assesment.csv")
Course_DF = pd.read_csv("D:\OneDrive\Desktop\Interview Prepe\Practical Exam\Tranning Analysis\courses.csv")

Asssesment_DF
```

```python
Course_DF
```

```python
print(f"Rows before remove Duplicates {len(Asssesment_DF)}")
Remove_Duplicates = Asssesment_DF.drop_duplicates()
print(f"Rows after remove Duplicates {len(Remove_Duplicates)}")
```

```python
Merged = Remove_Duplicates.merge(Course_DF , on = "course_id" , how = "left")
Merged
```

```python
Merged["Pass_Flag"] = (Merged["score"] >= 50).astype(int)
Merged
```

```python
Department_Summary = (Merged.groupby("department").agg(Total_Assesments = ("assessment_id" , "count") , Total_Passed_flags = ("Pass_Flag" , "sum")))
Department_Summary
```

```python
Department_Summary["Passed_Rate"] = (Department_Summary["Total_Passed_flags"] / Department_Summary["Total_Assesments"]) * 100
Department_Summary = Department_Summary.round(2)
Department_Summary
```

```python
Course_Summary = (Merged.groupby("course").agg(Total_Assesments = ("assessment_id" , "count") , Total_Passed_flags = ("Pass_Flag" , "sum")))
Course_Summary
```

```python
Course_Summary["Passed_Rate"] = (Course_Summary["Total_Passed_flags"] / Course_Summary["Total_Assesments"]) * 100
Course_Summary = Course_Summary.round(2)
Course_Summary
```

```python
Month_Order = ["Jan" , "Feb" , "Mar"]

Monthly_AVG = Merged.groupby("month")["score"].mean().reindex(["Jan" , "Feb" , "Mar"])


plt.figure(figsize = (8 , 5))
plt.plot(Monthly_AVG.index , Monthly_AVG.values , marker = "o")
plt.title("Monthly Average Assesment Score")
plt.xlabel("Months")
plt.ylabel("AVG. Score")

plt.show()
```
