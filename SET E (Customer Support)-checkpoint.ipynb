```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
Teams_DF = pd.read_csv("D:\\OneDrive\\Desktop\\Interview Prepe\\Practical Exam\\Customer Support\\teams.csv")
Tickets_DF = pd.read_csv("D:\\OneDrive\\Desktop\\Interview Prepe\\Practical Exam\\Customer Support\\tickets.csv")

Teams_DF
```

```python
Tickets_DF
```

```python
print(f"Rows before remove Duplicates {len(Tickets_DF)}")
Remove_Duplicates = Tickets_DF.drop_duplicates()
print(f"Rows after remove Duplicates {len(Remove_Duplicates)}")
```

```python
Merged = Tickets_DF.merge(Teams_DF , on = "team_id" , how = "left")
Merged
```

```python
Merged["Breach_Flag"] = (Merged["resolution_hours"] > 24).astype(int)
Merged
```

```python
Department_Summary = Merged.groupby("department").agg(Total_Tickets = ("ticket_id" , "count") , Total_Breach_Flag = ("Breach_Flag" , "sum"))

Department_Summary["sla_breach_rate"] = (Department_Summary["Total_Breach_Flag"] / Department_Summary["Total_Tickets"]) * 100
Department_Summary = Department_Summary.round(2)

Department_Summary
```

```python
Team_Summary = Merged.groupby(["team_id" , "team"]).agg(Total_Tickets = ("ticket_id" , "count") , 
                                                        Breach_Count = ("Breach_Flag" , "sum") ,
                                                        Avg_Revolution_Hours = ("resolution_hours" , "mean"))

Team_Summary["breach_rate"] = (Team_Summary["Breach_Count"] / Team_Summary["Total_Tickets"]) * 100
Team_Summary = Team_Summary.round(2)

Team_Summary
```

```python
Month_Order = ["Jan" , "Feb" , "Mar"]

Monthly_AVG = Merged.groupby("month")["resolution_hours"].mean().reindex(["Jan" , "Feb" , "Mar"])


plt.figure(figsize = (8 , 5))
plt.plot(Monthly_AVG.index , Monthly_AVG.values , marker = "o")
plt.title("Monthly Average Resolution Time")
plt.xlabel("Month")
plt.ylabel("Average Resolution Hours")

plt.show()
```
